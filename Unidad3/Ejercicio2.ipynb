```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 10, 20, 50])
p = np.array([0.50, 0.30, 0.15, 0.05])

# 1. Verificar probabilidades
print("Suma de probabilidades:", np.sum(p))

# 2. Distribucion de probabilidad
print("\nDistribucion de probabilidad")
for i in range(len(x)):
    print("X =", x[i], "P(X) =", p[i])

# 3. Valor esperado
esperado = np.sum(x * p)
print("\nValor esperado:", esperado)

# Varianza y desviacion estandar
varianza = np.sum((x - esperado)**2 * p)
desviacion = np.sqrt(varianza)

print("Varianza:", varianza)
print("Desviacion estandar:", desviacion)

# 4. Funcion acumulada
acumulada = np.cumsum(p)

print("\nFuncion de distribucion acumulada")
for i in range(len(x)):
    print("F(", x[i], ") =", acumulada[i])

# Grafica de probabilidad
plt.bar(x, p)
plt.xlabel("Premio (pesos)")
plt.ylabel("P(X)")
plt.title("Distribucion de probabilidad")
plt.xticks(x)
plt.grid(axis="y")
plt.show()

# Grafica acumulada
plt.step(x, acumulada, where="post")
plt.xlabel("Premio (pesos)")
plt.ylabel("F(x)")
plt.title("Funcion de distribucion acumulada")
plt.xticks(x)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.ylim(0, 1.1)
plt.grid()
plt.show()

```

##a) ¿Qué significa el valor esperado en este contexto?
Significa que el premio promedio a largo plazo es de $8.50 pesos por boleto, si se juega muchas veces bajo las probabilidades indicadas.
##b) ¿El juego es favorable para el jugador?

 Si cuesta más de $8.50, el valor esperado neto es negativo.
