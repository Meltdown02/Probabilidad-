```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3, 4, 5])
p = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])

# 1. Verificar probabilidades
print("Suma de probabilidades:", np.sum(p))

# 2. Valor esperado
esperado = np.sum(x * p)
print("Valor esperado:", esperado)

# 3. Varianza y desviacion estandar
varianza = np.sum((x - esperado)**2 * p)
desviacion = np.sqrt(varianza)

print("Varianza:", varianza)
print("Desviacion estandar:", desviacion)

# 4. Funcion de distribucion acumulada
acumulada = np.cumsum(p)
print("Funcion acumulada:", acumulada)

# Grafica de probabilidad
plt.bar(x, p)
plt.xlabel("Numero de celulas anormales")
plt.ylabel("P(X)")
plt.title("Distribucion de probabilidad")
plt.show()

# Grafica acumulada
plt.step(x, acumulada, where="post")
plt.xlabel("Numero de celulas anormales")
plt.ylabel("F(x)")
plt.title("Funcion de distribucion acumulada")
plt.ylim(0, 1.1)
plt.xticks(x)
plt.grid()
plt.show()

# 5. Interpretacion
print("Probabilidad de X <= 2:", acumulada[2] * 100, "%")
print("Probabilidad de X > 3:", np.sum(p[x > 3]) * 100, "%")
```

#Interpretación de los resultados en contexto biomédico
##a) ¿Cuál es el promedio esperado de células anormales por muestra?
El promedio esperado es de 2.1 células anormales por muestra, según la distribución de probabilidad proporcionada.

##b) ¿Existe mucha variabilidad entre las muestras?
La varianza es de 1.69 y la desviación estándar es de 1.30 células. Esto indica que existe cierta variación en la cantidad de células anormales entre las muestras, aunque los valores se concentran principalmente entre 1 y 3 células.
##c) ¿Qué indica la función acumulada sobre la probabilidad de encontrar un número bajo o alto de células anormales?
La función acumulada indica que la probabilidad de encontrar como máximo 2 células anormales es de 0.65, es decir, 65 %. En cambio, la probabilidad de encontrar más de 3 células es:

P(X>3)=P(X=4)+P(X=5)


P(X>3)=0.10+0.05=0.15


Por lo tanto, existe un 15 % de probabilidad de encontrar más de 3 células anormales en una muestra.
