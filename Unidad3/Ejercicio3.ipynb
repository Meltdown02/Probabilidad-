```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definir los datos del Ejercicio 3
# X: Unidades vendidas
x = np.array([0, 1, 2, 3, 4, 5])
# P(X): Probabilidad de cada valor
p = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

# Verificar que las probabilidades sumen 1
suma_prob = np.sum(p)
print(f"Suma de probabilidades: {suma_prob:.2f}")

# 2. Calcular el Valor Esperado E(X)
e_x = np.sum(x * p)
print(f"Valor Esperado E(X): {e_x:.2f}")

# 3. Calcular la Varianza y Desviación Estándar
e_x2 = np.sum((x**2) * p)
varianza = e_x2 - (e_x**2)
desv_est = np.sqrt(varianza)

print(f"Varianza (sigma^2): {varianza:.2f}")
print(f"Desviación Estándar (sigma): {desv_est:.2f}")

# 4. Calcular la Función de Distribución Acumulada F(x)
cdf = np.cumsum(p)

# 5. Generar las gráficas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Gráfica 1: Distribución de Probabilidad (PMF)
ax1.bar(
    x,
    p,
    color="skyblue",
    edgecolor="black",
    alpha=0.8,
    width=0.5,
    label="P(X)",
)
ax1.axvline(
    e_x,
    color="red",
    linestyle="dashed",
    linewidth=2,
    label=f"E(X) = {e_x}",
)
ax1.set_title("Distribución de Probabilidad de Ventas", fontsize=12, weight="bold")
ax1.set_xlabel("Unidades vendidas (X)", fontsize=10)
ax1.set_ylabel("Probabilidad P(X)", fontsize=10)
ax1.set_xticks(x)
ax1.set_ylim(0, 0.35)
ax1.grid(axis="y", linestyle="--", alpha=0.7)
ax1.legend()

# Gráfica 2: Función de Distribución Acumulada
x_step = np.append([-1, 0], x + 1)
cdf_step = np.append([0, 0], cdf)

ax2.step(
    x,
    cdf,
    where="post",
    color="darkgreen",
    linewidth=2,
    label="F(x) Acumulada",
)
ax2.plot(x, cdf, "go", markersize=6)
ax2.set_title("Función de Distribución Acumulada F(x)", fontsize=12, weight="bold")
ax2.set_xlabel("Unidades vendidas (X)", fontsize=10)
ax2.set_ylabel("Probabilidad Acumulada F(x)", fontsize=10)
ax2.set_xticks(x)
ax2.set_yticks(np.arange(0, 1.1, 0.2))
ax2.set_ylim(-0.05, 1.05)
ax2.grid(True, linestyle="--", alpha=0.7)
ax2.legend()

plt.tight_layout()
plt.show()
```
