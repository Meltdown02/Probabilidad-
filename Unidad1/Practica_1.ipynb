EJEMPLO 2: Tiempos de 100 metror de un grupo de 12

```python
#Definimos una variable llamada tiempos100mts de tipo lista que contiene los tiempos que tardaron los 10 alumnos para correr los 100 metros
tiempos100mts=[15,10,30,23,25,19,19,18,12,10]
print(tiempos100mts)
```

```python
print("El tiempo promedio de los corredores es de: ",mean(tiempos100mts))
```

```python
#Usamos la funcion median para determinar el tiempo promedio de la lista de valores
#Definimos una variable llamada mediana para calcular la mediana de la lista con la funcion median
mediana=median(tiempos100mts)
print("La media de los tiempos de los corredores es: ",mediana)
```

```python
#Definimos una variable llamada moda para calcular
multimodas=multimode(tiempos100mts)
print("La moda de los tiempos es: ",multimodas)
```

# Ejemplo 3: Histograma de 2 grupos
En este ejmplo generamos 100 numeros aleatorios reales que representan la produccion diaria de 2 fabricas de telas medidas en metros durante 100 dias
Generar otra lista con la produccion de tela de 100 dias para la segunda fabrica
Con estas -

```python
#Para generar numeros aleatorios
import random
```

```python
fabrica1=[]
for i in range(100):
    numeroaleatorio=round(random.uniform(50,100),2)
    fabrica1.append(numeroaleatorio)
print (fabrica1)
```

```python
fabrica2=[]
for i in range(100):
    numeroaleatorio=round(random.uniform(70,100),2)
    fabrica2.append(numeroaleatorio)
print (fabrica2)
```

```python
#Libreria de python para graficar valores
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("Produccion diara de la fabrica 1")
```

```python
plt.hist(fabrica2)
plt.title("Produccion diara de la fabrica 2")
```

```python
media1=mean(fabrica1)
mediana1=median(fabrica1)
moda1=mode(fabrica1)
print("Valores de la fabrica 1")
print("La media es ",media1)
print("La mediana es ",mediana1)
print("La moda es: ",moda1)
```

```python
media2=mean(fabrica2)
mediana2=median(fabrica2)
moda2=mode(fabrica2)
print("Valores de la fabrica 2")
print("La media es ",media2)
print("La mediana es ",mediana2)
print("La moda es: ",moda2)
```

```python
#Generar 2 subplot de 1 renglon por 2 columnas para desplegar los 2 histogramas una al lado de la otra
fig,(ax1, ax2)=plt.subplots(nrows=1,ncols=2,figsize=(10,5))
ax1.hist(fabrica1)
ax2.hist(fabrica2)
plt.show()
```

#Mostrar las medidas de tendencia central del histograma

```python
rangos=[50,55,60,65,70,75,80,85,90,100] #definir rangos de bins
plt.hist(fabrica1,bins=rangos,alpha=0.2,color='b',edgecolor="black")
plt.axvline(media1,color='b',linestyle='dashed',linewidth=2,label='Media')
plt.axvline(mediana1,color='r',linestyle='dashed',linewidth=2,label='Mediana')
plt.axvline(moda1,color='y',linestyle='dashed',linewidth=2,label='Moda')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Produccion diaria de la fabrica 1")
plt.legend()
plt.show()
```
