REALIZAR HISTOGRAMA BASADO EN LA REGLA DE STURGES

```python
n=len(autos.price)
k=int(1+math.log2(n))
print("Valor de n = ", n ,"\nNumero de intervalos =", k)
df = pd.read_csv("autos2.csv")
```

```python
plt.hist(df['price'], bins=k, alpha=0.5)
plt.xlabel('Precios')
plt.ylabel('Frecuencia')
plt.title('Histograma de precios basado en la regla de Sturges')
```

#CONCLUSIONES
La mayoria de precios son de 5000, despues de eso la frecuencia baja considerablemente a 10000, luego a 15000, y al llegar a 20000 la frecuencia es similar entre precios de 20000 a 45000

Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y tipo de vehículo (body style)

```python
sns.boxplot(x="price", y ="body-style", data=autos)
plt.xlabel('Precios')
plt.ylabel('Tipos de los vehiculos')
```

#CONCLUSIONES
Los precios de los convertibles van de 15000 a 30000, los de los hatchback de aproximadamente 8000 a 13000, de sedanes son de 9000 a 18000, los wagon van de 9000 a 15000 y los hardtop de 9500 a 33000

Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y tipo tracción ruedas (drive wheels)

```python
sns.boxplot(x="price", y ="drive-wheels", data=autos)
plt.xlabel('Precios')
plt.ylabel('Tipo de traccion ruedas')
```

#CONCLUSIONES
RWD (tracción trasera) presenta la mediana más alta y la mayor dispersión de precios. Además, tiene varios valores atípicos altos, lo que indica que incluye vehículos más costosos (posiblemente deportivos o de gama alta).

FWD (tracción delantera) tiene una mediana más baja y precios más concentrados en rangos económicos y medios. Presenta algunos valores atípicos, pero en general es el grupo más accesible.

4WD (tracción en las cuatro ruedas) muestra precios intermedios, con menor variabilidad que rwd. Tiene algún valor atípico, pero su rango es más compacto.

Realizar un diagrama de caja y bigotes que muestre la relación estadística entre la columna del precio del automóvil (price) y la ubicación del motor (engine-location)

```python

sns.boxplot(x="price", y ="engine-location", data=autos)
plt.xlabel('Precios')
plt.ylabel('Ubicacion del motor')
```

#CONCLUSIONES
##Motor delantero (front)
Presenta una mediana más baja  (10000–12000).
Tiene gran dispersión, lo que indica que existen modelos económicos y también algunos muy costosos.
Aparecen varios valores atípicos altos, lo que sugiere que algunos vehículos con motor delantero pueden alcanzar precios elevados.

##Motor trasero (rear)
Tiene una mediana mucho más alta (34000–35000).
La dispersión es mucho menor: los precios están más concentrados.
No se observan muchos valores atípicos, lo que indica mayor homogeneidad y concentracion de precios en esta categoría.
