```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generar datos para la distribución normal
media = 70
desv_std = 4
total_estudiantes = 500

peso_inferior = 68
peso_superior = 72

prob_menos_68kg = norm.cdf(peso_inferior, media, desv_std)
prob_menos_72kg = norm.cdf(peso_superior, media, desv_std)

prob_entre = prob_menos_72kg - prob_menos_68kg
estudiantes_entre = total_estudiantes * prob_entre

print("Probabilidad de que un estudiante pese más de 68 kg y menos de 72 kg:", prob_entre)
print("Estudiantes que pesan más de 68 kg y menos de 72 kg:", estudiantes_entre)

x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)

plt.fill_between(
    x,
    y,
    where=((x >= peso_inferior) & (x <= peso_superior)),
    color='skyblue',
    alpha=0.5
)

plt.title('Probabilidad de que un estudiante pese mas de 68 kg y menos de 72 kg')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)

plt.show()
```
